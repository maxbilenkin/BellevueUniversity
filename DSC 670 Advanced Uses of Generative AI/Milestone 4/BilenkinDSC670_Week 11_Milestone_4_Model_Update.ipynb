## Fine-Tuned GPT Model for Investment Research  

This project develops a fine-tuned GPT-3.5 model designed to generate structured investment analysis based on SEC 10-K (annual report) style prompts.  

The objective is to produce analyst-style summaries, risk assessments, and strategic insights to support investment decision-making.  

The final deliverable includes an interactive Streamlit application that allows users to submit financial prompts and receive real-time responses from the fine-tuned model.

## Project Overview

The objective of this project is to develop a fine-tuned GPT model capable of generating structured investment analysis from SEC 10-K filing prompts. 

Large language models can generate general financial explanations; however, fine-tuning improves response consistency, structure, and domain-specific reasoning. By training the model on curated investment-style prompts and responses, the system is optimized to produce:

- Executive-style company summaries  
- Identification of key financial and operational risks  
- Strategic comparisons between competitors  
- Clear, structured analyst-style outputs  

This approach enhances reliability and ensures responses align with professional investment research standards.

## Data Preparation

The training dataset was structured in JSONL format using prompt-response pairs compatible with OpenAI fine-tuning requirements.

Each training example follows a chat-based structure and includes:

- A **system message** defining the assistant as a professional investment analyst  
- A **user message** containing a financial research prompt (e.g., 10-K summary, risk analysis, strategic comparison)  
- An **assistant message** providing a structured analyst-style response  

The dataset was curated to ensure clarity, consistency, and professional tone. Responses were designed to emphasize financial performance, risk factors, strategic positioning, and long-term investment considerations.

## Training Dataset Creation

The following code constructs the structured training dataset in JSONL format. Each example follows the OpenAI chat fine-tuning schema and includes a system message, user prompt, and assistant response.

## Upload Training File

The following code uploads the JSONL training dataset to OpenAI in preparation for fine-tuning. 

The returned file ID is used to initiate the fine-tuning job in the next step.

```python
# Upload training file to OpenAI for fine-tuning
from openai import OpenAI
client = OpenAI()

file = client.files.create(
    file=open("investment_training_data.jsonl", "rb"),
    purpose="fine-tune"
)

print("File ID:", file.id)
```

### Test the Fine-Tuned Model

Before integrating the model into the Streamlit app, we test it using a few sample prompts to ensure it produces investment-focused summaries and risk analyses. This helps verify that the fine-tuning worked as expected.

```python
from openai import OpenAI

# Initialize OpenAI client
client = OpenAI()

# Fine-tuned model ID
fine_tuned_model_id = "ft:gpt-3.5-turbo-0125:personal::D8zPtSHE"

# Sample prompts
prompts = [
    "Summarize this 10-K filing for a long-term investor.",
    "Identify the top 5 risks mentioned in this 10-K filing.",
    "Compare this company's strategy to its main competitor."
]

# Generate outputs
for i, prompt in enumerate(prompts, start=1):
    response = client.chat.completions.create(
        model=fine_tuned_model_id,
        messages=[
            {"role": "system", "content": "You are an investment analyst."},
            {"role": "user", "content": prompt}
        ]
    )
    print(f"--- Output {i} ---")
    print(response.choices[0].message.content)
    print("\n")
```

### Build a Streamlit Application for Investment Research

This cell sets up a Streamlit interface to interact with the fine-tuned model. Users can input prompts related to SEC 10-K filings and receive analyst-style summaries, risk assessments, and strategic insights.

The application allows:

 - Input of custom prompts about financial filings.

 - Generation of responses using the fine-tuned investment research model.

 - Display of analyst-style output directly in the browser.

Run this cell using streamlit run app.py in a terminal to open the interactive app in your browser.

```python
import streamlit as st
from openai import OpenAI

# Initialize OpenAI client
client = OpenAI()

fine_tuned_model_id = "ft:gpt-3.5-turbo-0125:personal::D8zPtSHE"

st.title("Investment Research Assistant")
st.write("Enter prompts related to SEC 10-K filings and get analyst-style summaries and risk assessments.")

# User input
user_prompt = st.text_area("Enter your prompt here:", "")

if st.button("Generate Response"):
    if user_prompt.strip() == "":
        st.warning("Please enter a prompt before generating a response.")
    else:
        with st.spinner("Generating response..."):
            response = client.chat.completions.create(
                model=fine_tuned_model_id,
                messages=[
                    {"role": "system", "content": "You are an investment analyst."},
                    {"role": "user", "content": user_prompt}
                ]
            )
            output = response.choices[0].message.content
            st.subheader("Analyst Response")
            st.write(output)
```
